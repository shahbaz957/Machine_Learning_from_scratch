```python
import random
import numpy as np
```

```python
class KMeansCluster:
    def __init__(self , n_clusters = 2 , max_iter= 100):
        self.max_iter = max_iter
        self.n_clusters = n_clusters
        self.centroids = None

    def predict(self , X):
        random_index = random.sample(range(0,X.shape[0]) , self.n_clusters)
        self.centroids = X[random_index]
        for i in range (self.max_iter):
            cluster_group = self.assign_cluster(X)
            old_centroids = self.centroids

            self.centroids = self.move_centroids(X , cluster_group)
            if (old_centroids == self.centroids).all() : # as you are comparing two arrays so make sure you add .all()
                break

        return cluster_group
            
    def assign_cluster(self , X):
        cluster_group= []
        distances = []
        for row in X : 
            for centroid in self.centroids:
                distances.append(np.sqrt(np.dot(row-centroid , row-centroid)))
            min_distance = np.min(distances)
            min_idx = distances.index(min_distance)
            cluster_group.append(min_idx)
            distances.clear()
            
        return np.array(cluster_group)
    
    def move_centroids(self , X , cluster_group):
        new_centroid = []
        cluster_type = np.unique(cluster_group)
        for type in cluster_type:
            new_centroid.append(X[cluster_type == type].mean(axis = 0))

        return np.array(new_centroid)
        
```
