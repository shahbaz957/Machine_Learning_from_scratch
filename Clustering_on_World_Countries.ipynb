```python
df.drop(['Latitude' , 'Longitude'] , axis = 1 , inplace = True)
```

```python
cols = ['Density\n(P/Km2)','Agricultural Land( %)', 'Land Area(Km2)', 'Armed Forces size' , 'Co2-Emissions', 'Forested Area (%)' , 'Gasoline Price' , 'GDP' , 'Gross primary education enrollment (%)' , 'Gross tertiary education enrollment (%)' , 'Minimum wage' ,'Out of pocket health expenditure' , 'Population' ,'Population: Labor force participation (%)', 'Tax revenue (%)' , 'Total tax rate', 'Unemployment rate' , 'Urban_population' ]
df[cols] = (df[cols]
            .replace({',': '', '%': '', '$': ''}, regex=True)
            .apply(pd.to_numeric, errors='coerce'))
```

```python
df.info()
```

```python
df_numeric = df.select_dtypes(include=[np.integer , np.floating])
```

```python
df_numeric
```

```python
df_numeric.drop(['GDP', "Minimum wage"], axis = 1 , inplace = True)
```

```python
df_numeric
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df_numeric.isnull() , yticklabels=False)
plt.show()
```

```python
df_numeric = df_numeric.fillna(df_numeric.mean())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data_scaled = scaler.fit_transform(df_numeric)
```

```python
dff = pd.DataFrame(data_scaled)
```

```python
dff
```

```python

plt.figure(figsize=(12, 7))
plt.title("Country Dendrogram")
dend = shc.dendrogram(shc.linkage(data_scaled, method='ward'))
plt.show()
```

```python
cluster = AgglomerativeClustering(n_clusters=4, metric='euclidean', linkage='ward')
cluster_labels = cluster.fit_predict(data_scaled)

df['Cluster'] = cluster_labels
print(df['Cluster'].value_counts())
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(data_scaled[:,0], data_scaled[:,1], c=cluster_labels, cmap='tab10')
plt.title("Countries Clustered by HAC")
plt.show()
```

```python
df_result = pd.DataFrame({
    'Country': countries,
    'Cluster': cluster_labels
})
print(df_result.sort_values('Cluster'))
```

```python
from scipy.spatial.distance import pdist, squareform

distance_matrix = squareform(pdist(data_scaled, metric='euclidean'))
pd.DataFrame(distance_matrix) ## Proximity Matrix
```

```python
import numpy as np

for i, country in enumerate(countries):
    nearest_idx = np.argsort(distance_matrix[i])[1:4]  # skip itself, take 3 nearest
    nearest_countries = countries.iloc[nearest_idx].values
    print(f"{country} is closest to: {nearest_countries}")
```
