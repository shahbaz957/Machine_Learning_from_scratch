```python
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=90)
plt.show()
```

```python
def step(z):
    return 1 if z > 0 else 0

def perceptron(X , y):
    X = np.insert(X , 0 , 1 , axis = 1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    for i in range(1000):
        j = np.random.randint(0 , 100)
        y_hat = step(np.dot(X[j] , weights))
        weights = weights + (y[j] - y_hat)*X[j]
    return weights[0] , weights[1:]
    
```

```python
intercept_ , coef_ = perceptron(X, y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_ / coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m * x_input + b
```

```python
y_input
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

# **Logistic Regression Using SKlearn**

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(X,y)
```

```python
m = -(lor.coef_[0][0]/lor.coef_[0][1])
b = -(lor.intercept_/lor.coef_[0][1])
```

```python
x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.plot(x_input1,y_input1,color='black',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

### So the conclusion is that the logistic Regression class of sklearn is more optimized and good for classification purpose because once it classify the classes it further tries to optimize it self so that to reach the most optimized solution
