```python
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
import numpy as np
```

```python
customer_data = pd.read_csv("C:/Users/User/Downloads/hierarchical-clustering-with-python-and-scikit-learn-shopping-data.csv")
```

```python
customer_data
```

```python
data = customer_data.iloc[:,3:].values
```

```python
type(data)
```

```python
import scipy.cluster.hierarchy as shc
```

```python
plt.figure(figsize=(10, 7))
plt.title("Customer Dendograms")
dend = shc.dendrogram(shc.linkage(data , method = "ward"))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='ward')
labels_=cluster.fit_predict(data)
```

```python
pd.Series(labels_).value_counts()
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(data[:,0], data[:,1], c=cluster.labels_, cmap='magma')
plt.show()
```
